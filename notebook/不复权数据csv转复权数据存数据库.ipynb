```python
def to_tdx_code(code):
    bk = code.split("_")
    cd = code[0:6]
    if bk[1] == "XSHE":
        return "SZ." + cd
    elif bk[1] == "XSHG":
        return "SH." + cd
    else:
        print("不支持的股票市场")
        return code
df_feather.tail()
```
