```python
import pandas as pd
import tushare as ts
pro = ts.pro_api("f5ca7a922f41929faa97832f0fbe87e6ba523cf3e1db472612d3f3a4")
```

```python
def get_all_stocks_and_indexes():
    """
    获取所有A股股票和沪深指数代码
    
    Returns:
        股票和指数代码列表
    """
    
    try:
        # 获取所有A股股票
        stock_df = pro.stock_basic(
            exchange='',
            list_status='L',
            fields='ts_code,symbol,name,list_date,market'
        )
        
        # # 过滤掉ST、*ST、退市股票
        # stock_df = stock_df[
        #     ~stock_df['name'].str.contains('ST') &
        #     ~stock_df['name'].str.contains('退') &
        #     (stock_df['list_date'] <= '20251231')
        # ]
        
        # 获取沪深指数
        index_df = pro.index_basic(
            market='SSE,SZSE',
            fields='ts_code,name,list_date'
        )
        
        # 选择主要指数
        main_indexes = [
            '000001.SH',  # 上证指数
            '000300.SH',  # 沪深300
            '000905.SH',  # 中证500
            '399001.SZ',  # 深证成指
            '399006.SZ',  # 创业板指
            '399005.SZ',  # 中小板指
            '000016.SH',  # 上证50
            '000852.SH',  # 中证1000
        ]
        # index_df = index_df[index_df['ts_code'].isin(main_indexes)]
        print(len(stock_df), len(index_df))
        # 合并代码
        all_codes = pd.concat([
            stock_df['ts_code'],
            index_df['ts_code']
        ], ignore_index=True).drop_duplicates().tolist()
        
        print(f"获取完成: 共{len(stock_df)}只股票, {len(index_df)}个指数, 总计{len(all_codes)}个代码")
        return all_codes
        
    except Exception as e:
        print(f"获取代码列表失败: {e}")
        raise
```
